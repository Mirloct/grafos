```python
import networkx as nx

G = nx.from_pandas_edgelist(
    df_edgelist,
    source="txId1",
    target="txId2",
    create_using=nx.DiGraph()
)

print(G)
```

```python
import matplotlib.pyplot as plt

sub_nodes = list(G.nodes())[:80]

H = G.subgraph(sub_nodes)

plt.figure(figsize=(10,8))

nx.draw_networkx(
    H,
    node_size=40,
    arrows=True,
    with_labels=False
)

plt.show()
```

```python
print("Nodos:", G.number_of_nodes())
print("Aristas:", G.number_of_edges())
```

```python
in_degree = dict(G.in_degree())

out_degree = dict(G.out_degree())
```

```python
import pandas as pd

degree_df = pd.DataFrame({

    "txId": list(in_degree.keys()),
    "in_degree": list(in_degree.values()),
    "out_degree": list(out_degree.values())

})
```

## Módulo 5 y 6: Densidad y Caminos Mínimos

La **Densidad** mide qué tan cerca está el grafo de ser un 'grafo completo' (donde todos están conectados con todos). En redes de transacciones, suele ser muy baja.

El **Camino Mínimo** nos indica la distancia más corta entre dos transacciones, lo cual es vital para rastrear el flujo de fondos.

```python
density = nx.density(G)
print(f"Densidad de la red: {density:.6e}")

# Debido al tamaño del grafo, calculamos caminos mínimos sobre el subgrafo H
try:
    avg_path = nx.average_shortest_path_length(H)
    print(f"Camino mínimo promedio (subgrafo 80 nodos): {avg_path:.2f}")
except nx.NetworkXError:
    print("El subgrafo no es fuertemente conexo para calcular un promedio global.")
```

## Módulo 7: Métricas de Centralidad

Identificamos nodos influyentes mediante diversos algoritmos:
- **PageRank**: Mide la importancia basada en conexiones de calidad (usado por Google).
- **Betweenness**: Identifica nodos que actúan como 'puentes' entre diferentes partes de la red.

```python
# Calculamos PageRank para identificar nodos clave
pagerank = nx.pagerank(G, alpha=0.85)
degree_df['pagerank'] = degree_df['txId'].map(pagerank)

print("Top 5 nodos por PageRank:")
display(degree_df.sort_values('pagerank', ascending=False).head(5))
```

## Módulo 8: Detección de Comunidades

Utilizamos el algoritmo de **Comunidades de Louvain** (vía partición modular) para descubrir grupos de transacciones que operan de forma conjunta, lo cual es una señal clave para identificar clústeres de entidades o billeteras.

```python
from community import community_louvain

# Louvain requiere un grafo no dirigido
G_undirected = G.to_undirected()
partition = community_louvain.best_partition(G_undirected)

# Añadimos la comunidad al dataframe
degree_df['community'] = degree_df['txId'].map(partition)

print(f"Número de comunidades detectadas: {len(set(partition.values()))}")
display(degree_df['community'].value_counts().head(10))
```

## Módulo 9: Motifs y Patrones

Los 'motifs' son sub-estructuras recurrentes. En Bitcoin, un patrón de 'dispersión' (un nodo enviando a muchos) o 'agregación' (muchos enviando a uno) puede indicar mezcladores (tumblers) o exchanges.

```python
# Identificamos nodos con alto in-degree (posibles receptores/exchanges)
# y alto out-degree (posibles dispersores)
high_in = degree_df.nlargest(5, 'in_degree')
high_out = degree_df.nlargest(5, 'out_degree')

print("Posibles Hubs de Recepción (In-Degree):")
display(high_in)
print("\nPosibles Hubs de Dispersión (Out-Degree):")
display(high_out)
```

## Módulo 12: Introducción a Embeddings (SVD)

Para que un modelo de Deep Learning o ML entienda el grafo, convertimos la topología en vectores numéricos. Aquí usamos una aproximación basada en la descomposición de valores singulares (SVD) de la matriz de adyacencia.

```python
from scipy.sparse.linalg import svds
import numpy as np

# Obtenemos la matriz de adyacencia del grafo G
# Convertimos a float usando .astype(float) para compatibilidad con csr_array
A = nx.adjacency_matrix(G).astype(float)

# Reducción de dimensionalidad (SVD) para crear embeddings de 16 dimensiones
# k es el número de dimensiones deseadas
u, s, vt = svds(A, k=16)

print("Embeddings generados exitosamente.")
print(f"Forma de la matriz de embeddings (nodos, dimensiones): {u.shape}")
```
